```python


import pandas as pd
import numpy as np
from sklearn.datasets import load_iris
import matplotlib.pyplot as plt
import seaborn as sns

iris = load_iris()
df = pd.DataFrame(iris.data, columns = iris.feature_names)
df['species'] = iris.target
df['species'] = df['species'].map({0: 'Septosa', 1: 'Versicolor', 2: 'Virginica'})
print(df.head())
```

```python
print(df.describe())
```

```python
print("Mean Values: \n", df[iris.feature_names].mean())
```

```python
print("Median Values: \n", df[iris.feature_names].median())
```

```python
print("Mode Values: \n", df[iris.feature_names].mode())
```

```python
print("Minimum Values: \n", df[iris.feature_names].min())
```

```python
print("Maximum Values: \n", df[iris.feature_names].max())
```

```python
print("Standard Deviation Values: \n", df[iris.feature_names].std())
```

```python
# Graph Plot

# Histogram

df['sepal length (cm)'].hist(bins = 15, color = 'Magenta', edgecolor = 'black')
plt.title("Histogram of Sepal Length")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Frequency")
plt.show()
```

```python
# Box Plot

sns.boxplot(x = 'species', y = 'petal length (cm)', data = df)
plt.title("Box Plot of Petal Length by Species")
plt.ylabel("Petal Length (cm)")
plt.show()
```

```python
# Scatter Plot (Sepal vs Petal Length)

sns.scatterplot(x = 'sepal length (cm)', y = 'petal length (cm)', hue = 'species', data = df)
plt.title("Scatter plot of Sepal vs Petal Length")
plt.show()
```

```python
# Experiment 2

df_csv = pd.read_csv('students_data.csv')
print(" Student Performance: \n", df_csv.head())
```

```python
# Descriptive Statistics

print("Descriptive Statistics: \n",df_csv.describe())
```

```python
# Data Distribution Characterization

print("Data distribution characteristics:")
numerical_cols = df_csv.select_dtypes(include=np.number)

print("\nSkewness:")
print(numerical_cols.skew())

print("\nKurtosis:")
print(numerical_cols.kurt())
```

```python
df_csv[['Age', 'ApproxHeight', 'ApproxWeight']].skew()
```

```python
df_csv[['Age', 'ApproxHeight', 'ApproxWeight']].kurtosis()
```

```python
# Graphical Visualization Techniques

# LinePlot

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.lineplot(data = df_csv, x = "ApproxHeight", y = "ApproxWeight", marker = "o")
plt.title("Line Plot of ApproxHeight vs ApproxWeight")
plt.xlabel("ApproxHeight")
plt.ylabel("ApproxWeight")
plt.show()
```

```python
# Outlier Detection Using Visual Methods

# Histogram

df_csv['ApproxHeight'].hist(bins = 15, color = 'Yellow', edgecolor = 'black')
plt.title("Histogram of Approximate Height")
plt.xlabel("Approximate Height")
plt.ylabel("Frequency")
plt.show()
```

```python
# Box Plot

sns.boxplot(x = 'Age', y = 'ApproxWeight', data = df_csv)
plt.title("Box Plot of Approximate Weight by Age")
plt.ylabel("ApproxWeight")
plt.show()
```

```python
# Scatter Plot

sns.scatterplot(x = 'ApproxHeight', y = 'ApproxWeight', hue = 'Age', data = df_csv)
plt.title("Scatter plot of ApproxHeight vs ApproxWeight")
plt.show()
```
