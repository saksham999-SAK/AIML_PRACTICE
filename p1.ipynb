```python
import matplotlib.pyplot as plt
```

```python
plt.title('Distribution of Math Scores')
plt.xlabel('Math Score')
plt.ylabel('Frequency')
plt.hist(df['math score'])
plt.show()
```

```python
plt.title('Distribution of reading scores')
plt.xlabel('Reading Score')
plt.ylabel('Frequency')
plt.hist(df['reading score'])
plt.show()
```

```python
plt.title('Distribution of writing scores')
plt.xlabel('writing Score')
plt.ylabel('Frequency')
plt.hist(df['writing score'])
plt.show()
```

```python
import seaborn as sns
sns.countplot(data=df, x='gender')
plt.title('Distribution of Gender')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

```python
race_counts = df['race/ethnicity'].value_counts()
plt.figure(figsize=(7, 7))
plt.pie(race_counts, labels=race_counts.index, autopct='%1.1f%%')
plt.title('Distribution of race/ethinicity')
plt.show()
```

```python
plt.figure(figsize=(10, 8))
sns.scatterplot(data=df, x='math score', y='reading score', hue='gender')
plt.title('Math Score vs. Reading Score')
plt.xlabel('Math Score')
plt.ylabel('Reading Score')
plt.show()
```

```python
level_counts = df['parental level of education'].value_counts()
plt.figure(figsize=(7, 7))
plt.pie(level_counts, labels=level_counts.index, autopct='%1.1f%%')
plt.title('Distribution of parental level of education')
plt.show()
```
