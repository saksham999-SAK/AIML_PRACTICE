```python
# 2. Identifying Types of Data
# Write a Python program to :
# Read a dataset containing student information

df_csv = pd.read_csv("students_data.csv")
print(df_csv)
```

```python
# Identify and separate:
# Numerical columns
# Categorical columns

numerical_columns = df_csv.select_dtypes(include = ['int64', 'float64']).columns
categorical_columns = df_csv.select_dtypes(include = ['object']).columns

print("Numerical Columns: \n")
print(numerical_columns)
print("\n Categorical Columns: \n")
print(categorical_columns)
```

```python
ordinal_columns = ['YearofStudy', 'Age', 'Course']
ordinal_columns = [col for col in ordinal_columns if col in df_csv.columns]
print("Ordinal Columns: ", ordinal_columns)
```
